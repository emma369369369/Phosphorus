```python
# ==============================================================================
# SELF-IMPROVING LLM: STEP 1 - DATA GENERATION (THE TEACHER)
# Copy and run this entire script in a Google Colab Cell (T4 GPU).
# ==============================================================================

import subprocess
import sys
import os

print("--- CHECKING ENVIRONMENT ---")

# 1. INSTALL DEPENDENCIES VERBOSELY
# We run this with subprocess to ensure it finishes before Python tries to import anything.
print("Installing dependencies (torch, transformers, bitsandbytes)... plese wait.")
subprocess.check_call([sys.executable, "-m", "pip", "install", "-U", "torch", "transformers", "accelerate", "bitsandbytes"])
print("Dependencies installed.")

# 2. CHECK FOR GPU
import torch
if not torch.cuda.is_available():
    raise RuntimeError(
        "\n\nCRITICAL ERROR: GPU not detected.\n"
        "You are running on a CPU Runtime.\n"
        "Please go to the top menu: Runtime > Change runtime type > Select 'T4 GPU' > Save.\n"
        "Then try running this cell again."
    )
else:
    print(f"GPU Detected: {torch.cuda.get_device_name(0)}")

# 3. IMPORTS (Must happen after install)
import json
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig

# --- CONFIGURATION ---
MODEL_ID = "mistralai/Mistral-7B-v0.1"
OUTPUT_FILE = "synthetic_code_dataset.jsonl"
NUM_EXAMPLES_PER_TOPIC = 5

# Topics we want the model to generate code for
TOPICS = [
    "file manipulation",
    "network socket programming",
    "data encryption using fernet",
    "web scraping with beautifulsoup",
    "system process management",
    "multithreading and concurrency",
    "analyzing log files using regex"
]

# --- MODEL SETUP ---
print(f"Loading {MODEL_ID}...")
try:
    bnb_config = BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_quant_type="nf4",
        bnb_4bit_compute_dtype=torch.bfloat16
    )

    tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)
    model = AutoModelForCausalLM.from_pretrained(
        MODEL_ID,
        quantization_config=bnb_config,
        device_map="auto"
    )
except ImportError as e:
    print(f"\n\nIMPORT ERROR: {e}")
    print("If you see an error about 'bitsandbytes', keep the GPU runtime active,")
    print("and click 'Runtime > Restart Session' in the top menu, then run this cell again.")
    raise e

# --- THE TEACHER PROMPT (FEW-SHOT) ---
# This serves as the "System Prompt" for the Base Model.
# We show it examples of what we want, so it continues the pattern.
TEACHER_PROMPT = """
# Python Code Generator
# This AI generates high-quality, executable Python scripts for specific tasks.

### Task: Calculate the factorial of a number
### Code:
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

if __name__ == "__main__":
    print(factorial(5))

### Task: Check if a number is prime
### Code:
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

### Task: {TOPIC}
### Code:
"""

# --- GENERATION FUNCTION ---
def generate_code(topic):
    prompt = TEACHER_PROMPT.format(TOPIC=topic)
    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")

    outputs = model.generate(
        **inputs,
        max_new_tokens=256,   # Limit length to keep it focused
        do_sample=True,       # Add creativity
        temperature=0.7,
        top_p=0.9,
        pad_token_id=tokenizer.eos_token_id,
        stop_strings=["### Task:"], # Stop it if it tries to start a new task immediately
        tokenizer=tokenizer
    )

    # Extract only the newly generated text
    full_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
    generated_code = full_text.split(prompt)[1].strip()

    # Simple cleanup: stop at the next "### Task" if the specific stop token didn't catch it
    if "### Task:" in generated_code:
        generated_code = generated_code.split("### Task:")[0]

    return generated_code

# --- MAIN LOOP ---
data = []
print(f"Starting generation for {len(TOPICS)} topics...")

for topic in TOPICS:
    print(f"\n--> Generating examples for: {topic}")
    for i in range(NUM_EXAMPLES_PER_TOPIC):
        try:
            code = generate_code(topic)

            # Save strictly in the format needed for Fine-Tuning later
            entry = {
                "instruction": f"Write a Python script for: {topic}",
                "output": code
            }
            data.append(entry)
            print(f"   [v] Generated example {i+1}")

        except Exception as e:
            print(f"   [x] Error: {e}")

# --- SAVE TO FILE ---
print(f"\nSaving {len(data)} examples to {OUTPUT_FILE}...")
with open(OUTPUT_FILE, 'w') as f:
    for entry in data:
        f.write(json.dumps(entry) + "\n")

print("Done! Download the .jsonl file from the files tab.")
```
